```python
# 数据增强函数：简单的 SMILES 序列翻转
def augment_smiles(smiles):
    """简单的数据增强方法，例如旋转 SMILES 字符串"""
    if random.random() > 0.5:
        return smiles[::-1]  # 翻转字符串
    return smiles

# 根据 mgperL 浓度生成分类标签
def generate_labels(mgperL):
    """根据 mgperL 的浓度范围生成分类标签"""
    if mgperL < 0.2:
        return 0  # high
    elif 0.2 <= mgperL < 1.5:
        return 1  # mid
    elif 1.5 <= mgperL < 3.5:
        return 2  # mid
    else:
        return 3  # low

# 数据集定义
class SMILES_Dataset(Dataset):
    def __init__(self, smiles, reg_labels, class_labels):
        self.smiles = smiles
        self.reg_labels = reg_labels  # 回归任务标签 (mgperL)
        self.class_labels = class_labels  # 分类任务标签

    def __len__(self):
        return len(self.smiles)

    def __getitem__(self, idx):
        #smiles = augment_smiles(self.smiles[idx])  # 数据增强
        smiles = self.smiles[idx]  # 数据增强
        reg_label = self.reg_labels[idx]
        class_label = self.class_labels[idx]
        tokens = tokenizer(smiles, padding='max_length', truncation=True, max_length=128, return_tensors="pt")
        return tokens, torch.tensor(reg_label, dtype=torch.float32), torch.tensor(class_label, dtype=torch.long)

# 多任务模型定义
class ChemBERTa_MultiTask(nn.Module):
    def __init__(self, num_classes):
        super(ChemBERTa_MultiTask, self).__init__()
        # 加载预训练的ChemBERTa模型
        self.chemberta = chemberta_model
        hidden_size = self.chemberta.config.hidden_size  # 一般为768
        # 回归任务的全连接层
        self.regressor = nn.Sequential(
            nn.Linear(hidden_size, 512),
            nn.ReLU(),
            nn.BatchNorm1d(512),
            nn.Dropout(0.3),
            nn.Linear(512, 256),
            nn.ReLU(),
            nn.BatchNorm1d(256),
            nn.Dropout(0.3),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.BatchNorm1d(128),
            nn.Dropout(0.3),
            nn.Linear(128, 1)
        )

        # 分类任务的全连接层
        self.classifier = nn.Sequential(
            nn.Linear(hidden_size, 512),
            nn.ReLU(),
            nn.BatchNorm1d(512),
            nn.Dropout(0.3),
            nn.Linear(512, 256),
            nn.ReLU(),
            nn.BatchNorm1d(256),
            nn.Dropout(0.3),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.BatchNorm1d(128),
            nn.Dropout(0.3),
            nn.Linear(128, 4)
        )

    def forward(self, tokens):
        output = self.chemberta(**tokens)
        cls_embedding = output.last_hidden_state[:, 0, :]  # [CLS] token 嵌入
        reg_output = self.regressor(cls_embedding)  # 回归任务输出
        class_output = self.classifier(cls_embedding)  # 分类任务输出
        return reg_output, class_output
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from rdkit import Chem
from rdkit.Chem import Draw
from collections import defaultdict

# 确保 ChemBERTa 模型输出 attentions
chemberta_model.config.output_attentions = True

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)
model.eval()

# 将 Token-level attention 映射到 atom-level
def map_token_attention_to_atoms(smiles, token_attention, tokens):
    mol = Chem.MolFromSmiles(smiles)
    atom_attn = np.zeros(mol.GetNumAtoms(), dtype=float)
    counts = np.zeros(mol.GetNumAtoms(), dtype=float)
    # 获取每个 atom 对应的 SMILES 片段
    atom_fragments = [
        Chem.MolFragmentToSmiles(mol, [i], canonical=True) 
        for i in range(mol.GetNumAtoms())
    ]
    for atom_idx, frag in enumerate(atom_fragments):
        for tok_idx, tok in enumerate(tokens):
            if frag in tok:
                atom_attn[atom_idx] += token_attention[tok_idx]
                counts[atom_idx] += 1
    counts[counts == 0] = 1
    return atom_attn / counts

# 提取一系列 SMILES 的 atom-level attention
def extract_attention_weights(smiles_list, model, tokenizer, layer_idx=-1):
    all_atom_attn = []
    for smi in smiles_list:
        # 编码并移动到 device
        inputs = tokenizer(smi, return_tensors="pt", padding=True,
                           truncation=True, max_length=128)
        inputs = {k: v.to(device) for k, v in inputs.items()}
        with torch.no_grad():
            outputs = model.chemberta(**inputs)
        attentions = outputs.attentions  # tuple of [batch, heads, seq, seq]

        # 选择指定层并平均所有 head
        layer_attn = attentions[layer_idx].mean(dim=1)[0]  # [seq, seq]

        # 获取 input_ids 并转为 tokens（排除首尾特殊符号）
        input_ids = inputs['input_ids'][0]
        tokens_all = tokenizer.convert_ids_to_tokens(input_ids)
        tokens = tokens_all[1:len(tokens_all)-1]

        # 从 [CLS] (idx=0) 到每个实际 token
        seq_len = len(tokens)
        cls_to_tokens = layer_attn[0, 1:1+seq_len].cpu().numpy()

        # 映射到 atom
        atom_attn = map_token_attention_to_atoms(smi, cls_to_tokens, tokens)
        # 归一化
        atom_attn = (atom_attn - atom_attn.min()) / (atom_attn.max() - atom_attn.min() + 1e-8)
        all_atom_attn.append(atom_attn)
    return all_atom_attn

# 分子可视化：在 RDKit 二维图上按 attention 着色
def visualize_molecule_attention(smiles, atom_attention):
    mol = Chem.MolFromSmiles(smiles)
    atom_colors = {
        i: (1.0, 1.0 - atom_attention[i], 1.0 - atom_attention[i]) 
        for i in range(len(atom_attention))
    }
    drawer = Draw.MolDraw2DCairo(400, 400)
    Draw.rdMolDraw2D.PrepareAndDrawMolecule(
        drawer, mol,
        highlightAtoms=list(atom_colors.keys()),
        highlightAtomColors=atom_colors
    )
    drawer.FinishDrawing()
    png = drawer.GetDrawingText()
    from IPython.display import Image, display
    display(Image(png))

# 定义常见官能团 SMARTS
functional_groups = {
    "Hydroxyl": "[OX2H]", 
    "Carbonyl": "[CX3]=O", 
    "Amine": "[NX3;H2,H1;!$(NC=O)]",
    "Carboxyl": "[CX3](=O)[OX2H1]",
    "Halogen": "[F,Cl,Br,I]"
}

# 统计官能团 average attention
def compute_functional_group_attention(smiles_list, all_attentions):
    fg_scores = defaultdict(list)
    for smi, attn in zip(smiles_list, all_attentions):
        mol = Chem.MolFromSmiles(smi)
        for fg_name, fg_smarts in functional_groups.items():
            patt = Chem.MolFromSmarts(fg_smarts)
            for atom in mol.GetAtoms():
                if atom.HasSubstructMatch(patt):
                    fg_scores[fg_name].append(attn[atom.GetIdx()])
    return {fg: float(np.mean(scores)) for fg, scores in fg_scores.items() if scores}

# 绘制官能团重要性柱状图
def plot_functional_group_importance(fg_avg):
    names = list(fg_avg.keys())
    values = [fg_avg[n] for n in names]
    plt.figure(figsize=(6,4))
    plt.bar(names, values)
    plt.ylabel("Average Attention Score")
    plt.title("Functional Group Importance (Toxicity Task)")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


```

```python
# -*- coding: utf-8 -*-
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
from rdkit import Chem, RDLogger
from rdkit.Chem import Draw
from transformers import RobertaTokenizer, RobertaModel
from collections import defaultdict
import pandas as pd

# ----------------------------
# 1. 准备：加载数据 & 关闭 RDKit 日志
# ----------------------------
RDLogger.DisableLog('rdApp.*')



# ----------------------------
# 3. 工具函数：Token-level → Atom-level
# ----------------------------
def map_token_attention_to_atoms(smiles, token_attn, tokens):
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return []
    n = mol.GetNumAtoms()
    atom_attn = np.zeros(n, dtype=float)
    counts    = np.zeros(n, dtype=float)
    frags = [Chem.MolFragmentToSmiles(mol, [i], canonical=True) 
             for i in range(n)]
    for i, frag in enumerate(frags):
        for j, tok in enumerate(tokens):
            if frag in tok:
                atom_attn[i] += token_attn[j]
                counts[i]    += 1
    counts[counts==0] = 1
    return (atom_attn / counts).tolist()

# ----------------------------
# 4. 提取 Atom-level Attention
# ----------------------------
def extract_attention_weights(smiles_list, model, tokenizer, layer_idx=-1):
    valid_smiles = []
    all_attn     = []
    for smi in smiles_list:
        if Chem.MolFromSmiles(smi) is None:
            continue

        # tokenize & forward
        inputs = tokenizer(
            smi, return_tensors="pt",
            padding=True, truncation=True, max_length=128
        )
        inputs = {k: v.to(device) for k, v in inputs.items()}
        with torch.no_grad():
            out = model.chemberta(**inputs)
        atts = out.attentions[layer_idx].mean(dim=1)[0]  # [seq,seq]

        # 拿到实际 tokens (去掉 [CLS] & [SEP])
        toks_all = tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])
        toks     = toks_all[1:-1]
        seq_len  = len(toks)
        cls2tok  = atts[0, 1:1+seq_len].cpu().numpy()

        atom_attn = map_token_attention_to_atoms(smi, cls2tok, toks)
        if not atom_attn:
            continue
        arr = np.array(atom_attn)
        # 归一化到 [0,1]
        arr = (arr - arr.min()) / (arr.max() - arr.min() + 1e-8)

        valid_smiles.append(smi)
        all_attn.append(arr.tolist())

    return valid_smiles, all_attn

# ----------------------------
# 5. 可视化单分子 Attention
# ----------------------------
def visualize_molecule_attention(smiles, atom_attention):
    mol = Chem.MolFromSmiles(smiles)
    atom_colors = {
        i: (1.0, 1.0 - atom_attention[i], 1.0 - atom_attention[i])
        for i in range(len(atom_attention))
    }
    drawer = Draw.MolDraw2DCairo(400, 400)
    Draw.rdMolDraw2D.PrepareAndDrawMolecule(
        drawer, mol,
        highlightAtoms=list(atom_colors.keys()),
        highlightAtomColors=atom_colors
    )
    drawer.FinishDrawing()
    png = drawer.GetDrawingText()
    from IPython.display import Image, display
    display(Image(png))

# ----------------------------
# 6. 官能团 & 原子类型 统计分析
# ----------------------------
# 定义官能团 SMARTS
functional_groups = {
    "Hydroxyl":  "[OX2H]",
    "Carbonyl":  "[CX3]=O",
    "Amine":     "[NX3;H2,H1;!$(NC=O)]",
    "Carboxyl":  "[CX3](=O)[OX2H1]",
    "Halogen":   "[F,Cl,Br,I]"
}

def compute_and_plot_fg_importance(smiles_list, all_attn):
    fg_scores = defaultdict(list)
    for smi, attn in zip(smiles_list, all_attn):
        mol = Chem.MolFromSmiles(smi)
        for name, smarts in functional_groups.items():
            patt = Chem.MolFromSmarts(smarts)
            for match in mol.GetSubstructMatches(patt):
                for idx in match:
                    fg_scores[name].append(attn[idx])
    names, vals = [], []
    for name, scores in fg_scores.items():
        if scores:
            names.append(name)
            vals.append(float(np.mean(scores)))
    plt.figure(figsize=(6,4))
    plt.bar(names, vals)
    plt.ylabel("Avg Attention")
    plt.title("Functional Group Importance")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

def compute_and_plot_atom_type_importance(smiles_list, all_attn):
    at_scores = defaultdict(list)
    for smi, attn in zip(smiles_list, all_attn):
        mol = Chem.MolFromSmiles(smi)
        for atom in mol.GetAtoms():
            at_scores[atom.GetSymbol()].append(attn[atom.GetIdx()])
    names, vals = [], []
    for elem, scores in at_scores.items():
        if scores:
            names.append(elem)
            vals.append(float(np.mean(scores)))
    plt.figure(figsize=(6,4))
    plt.bar(names, vals)
    plt.ylabel("Avg Attention")
    plt.title("Atom Type Importance")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# ----------------------------
# 7. 示例调用（注释掉后自行启用）
# ----------------------------


```

```python
import matplotlib.pyplot as plt
import numpy as np
from rdkit import Chem
from rdkit.Chem import Draw
from PIL import Image
import io

def draw_molecule_attention_highres(
    smiles: str,
    atom_attn: list,
    canvas_size: tuple = (600, 600),
    dpi: int = 300,
    colormap: str = 'hot',
    highlight_radius: float = 0.6,
    bond_line_width: float = 2.0
):
    """
    在高分辨率下绘制分子原子与键的 Attention 热度图，
    并加粗 bond 线条，将 atom label 文字设为黑色。
    """
    # 1) 解析分子，计算 bond attention
    mol = Chem.MolFromSmiles(smiles)
    n_atoms = mol.GetNumAtoms()
    # 计算 bond-level attention 为两端 atom attention 的平均
    bond_attn = {}
    for bond in mol.GetBonds():
        i = bond.GetBeginAtomIdx()
        j = bond.GetEndAtomIdx()
        bond_idx = bond.GetIdx()
        bond_attn[bond_idx] = float((atom_attn[i] + atom_attn[j]) / 2.0)

    # 2) RDKit 绘图到 PNG bytes
    atom_colors = {idx: plt.get_cmap(colormap)(att)[:3]
                   for idx, att in enumerate(atom_attn)}
    bond_colors = {idx: plt.get_cmap(colormap)(att)[:3]
                   for idx, att in bond_attn.items()}

    drawer = Draw.MolDraw2DCairo(canvas_size[0], canvas_size[1])
    opts = drawer.drawOptions()
    # 增加 bond 粗细
    opts.bondLineWidth = bond_line_width
    # 白底、少 padding
    opts.backgroundColour = (1,1,1)
    opts.padding = 0.05
    # 强制 atom label 黑色
    opts.atomLabelFontSize = 18
    opts.atomLabelColour = (0,0,0)

    Draw.rdMolDraw2D.PrepareAndDrawMolecule(
        drawer, mol,
        highlightAtoms=list(atom_colors.keys()),
        highlightAtomColors=atom_colors,
        highlightAtomRadii={i: highlight_radius for i in atom_colors},
        highlightBonds=list(bond_colors.keys()),
        highlightBondColors=bond_colors
    )
    drawer.FinishDrawing()
    png = drawer.GetDrawingText()
    pil_img = Image.open(io.BytesIO(png))

    # 3) Matplotlib 排版 + 水平 colorbar
    fig = plt.figure(
        figsize=(canvas_size[0]/dpi, canvas_size[1]/dpi),
        dpi=dpi
    )
    gs = fig.add_gridspec(2, 1, height_ratios=[1, 0.05], hspace=0.02)
    ax_mol = fig.add_subplot(gs[0, 0])
    ax_bar = fig.add_subplot(gs[1, 0])

    ax_mol.imshow(pil_img)
    ax_mol.axis('off')

    norm = plt.Normalize(vmin=0, vmax=1)
    sm = plt.cm.ScalarMappable(cmap=colormap, norm=norm)
    sm.set_array([])

    cbar = fig.colorbar(
        sm, cax=ax_bar, orientation='horizontal',
        ticks=[0.0, 0.5, 1.0]
    )
    cbar.set_label("Attention score", fontsize=8)
    cbar.ax.tick_params(labelsize=6)

    return fig, ax_mol


fig, ax = draw_molecule_attention_highres(
    valid_smiles[0],
    all_attn[0],
    canvas_size=(600,600),
    dpi=300,
    colormap='Blues',
    highlight_radius=0.6
)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA



smiles_embeddings = np.load('./fish_unique_smiles_embeddings.npy')


data=pd.read_excel("./fish_unique.xlsx")
data=data.dropna()

mgperL = data['mgperL'].values
mgperL=np.log1p(mgperL)



# 根据 mgperL 浓度生成分类标签
def generate_labels(mgperL):
    """根据 mgperL 的浓度范围生成分类标签"""
    if mgperL < 0.2:
        return 0  # high
    elif 0.2 <= mgperL < 1.5:
        return 1  # mid
    elif 1.5 <= mgperL < 3.5:
        return 2  # mid
    else:
        return 3  # low
# 生成分类标签    
classification_labels = [generate_labels(mg) for mg in mgperL]


import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.manifold import TSNE
from sklearn.metrics import davies_bouldin_score
import pandas as pd

# 假设 smiles_embeddings 是 [N, 768] 的词嵌入，classification_labels 是 [N] 的分类标签 (如 0,1,2,3)

# 使用 t-SNE 降维
tsne = TSNE(n_components=2, random_state=42, perplexity=30, n_iter=1000)
tsne_result = tsne.fit_transform(smiles_embeddings)

# 直接使用离散标签作为分组
dbi_tsne = davies_bouldin_score(tsne_result, classification_labels)

```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.random_projection import GaussianRandomProjection
from scipy.stats import gaussian_kde
from matplotlib.colors import Normalize
import matplotlib.cm as cm
import seaborn as sns





# 使用 Seaborn 的 "ticks" 风格，确保刻度线和刻度显示
sns.set_theme(context="notebook", style="ticks")

# ========================
# 1. 加载你的 ChemBERTa 嵌入数据
# ========================
embeddings = np.load("fish_unique_smiles_embeddings.npy")
print("Embeddings shape:", embeddings.shape)

# ========================
# 2. 随机投影 + 单位化
# ========================
# (a) L2 归一化
emb_norm = embeddings / np.linalg.norm(embeddings, axis=1, keepdims=True)
# (b) 随机正交投影到 2D
projector = GaussianRandomProjection(n_components=2, random_state=42)
proj_2d = projector.fit_transform(emb_norm)
# (c) 再次归一化，让所有点都落在单位圆上
proj_2d /= np.linalg.norm(proj_2d, axis=1, keepdims=True)
# (d) 计算每个点的极角 ∈ [-π, π]
angles = np.arctan2(proj_2d[:, 1], proj_2d[:, 0])

# ========================
# 3. 计算 2D KDE 并在圆周上采样
# ========================
kde_2d = gaussian_kde(proj_2d.T, bw_method=0.1)
theta_vals = np.linspace(-np.pi, np.pi, 360, endpoint=False)
ring_x = np.cos(theta_vals)
ring_y = np.sin(theta_vals)
densities = kde_2d(np.vstack([ring_x, ring_y]))
dens_min, dens_max = densities.min(), densities.max()
norm = Normalize(vmin=dens_min, vmax=dens_max)
cmap = cm.get_cmap("RdPu")

# ========================
# 4. 角度分布的 1D KDE
# ========================
kde_1d = gaussian_kde(angles, bw_method=0.3)
x_angle = np.linspace(-np.pi, np.pi, 400)
y_angle = kde_1d(x_angle)

# ========================
# 5. 绘图（dpi=500）
# ========================
fig, (ax_top, ax_bottom) = plt.subplots(
    2, 1,
    figsize=(6, 10),
    gridspec_kw={"height_ratios": [3, 1], "hspace": 0.25},
    dpi=500
)

# ~~~~~~~~~~~~ (A) 上图：彩色环 ~~~~~~~~~~~~
line_width = 20
for i in range(len(theta_vals) - 1):
    x1, y1 = ring_x[i], ring_y[i]
    x2, y2 = ring_x[i+1], ring_y[i+1]
    density_i = densities[i]
    color = cmap(norm(density_i))
    ax_top.plot([x1, x2], [y1, y2], color=color, linewidth=line_width, solid_capstyle='round')

# 叠加虚线圆边框（可选，根据需要开启或关闭）
circle_outline = plt.Circle((0, 0), 1.0, color='black', fill=False, lw=1.5, linestyle='--')
ax_top.add_patch(circle_outline)

# 调整上图坐标与刻度设置
ax_top.set_aspect("equal", "box")
ax_top.set_xlim(-1.1, 1.1)
ax_top.set_ylim(-1.1, 1.1)
ax_top.set_xticks([-1, -0.5, 0, 0.5, 1])
ax_top.set_yticks([-1, -0.5, 0, 0.5, 1])
for spine in ax_top.spines.values():
    spine.set_linewidth(1.5)

# 设置刻度线样式
ax_top.tick_params(which='both', direction='out', length=5, width=1.5, labelsize=25)
# 加粗刻度标签（数字）
for label in ax_top.get_xticklabels() + ax_top.get_yticklabels():
    label.set_fontweight('bold')

# 加粗坐标轴标签
ax_top.set_ylabel("Features", fontsize=30, fontweight='bold')
ax_top.set_xlabel("Features", fontsize=30, fontweight='bold')

# ============ (B) 下图设置 ============
ax_bottom.plot(x_angle, y_angle, color="steelblue", lw=2)
ax_bottom.fill_between(x_angle, y_angle, alpha=0.3, color="steelblue")
ax_bottom.set_xlim(-np.pi, np.pi)
ax_bottom.tick_params(which='both', direction='out', length=5, width=1.5, labelsize=25)
# 加粗刻度标签
for label in ax_bottom.get_xticklabels() + ax_bottom.get_yticklabels():
    label.set_fontweight('bold')

# 加粗坐标轴标签
ax_bottom.set_xlabel("Angles", fontsize=30, fontweight='bold')
ax_bottom.set_ylabel("Density", fontsize=30, fontweight='bold')

plt.tight_layout()
plt.show()
```
